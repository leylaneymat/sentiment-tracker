```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
raw_data = pd.read_csv('../data/raw_data/twitter_data.csv', encoding_errors='ignore', header=None)
```

```python
raw_data.columns = ["target", "ids", "date", "flag", "user", "text"]
```

```python
raw_data
```

### About the data
- target: the polarity of the tweet (0 = negative, 2 = neutral, 4 = positive)
- ids: The id of the tweet (2087)
- date: the date of the tweet (Sat May 16 23:58:44 UTC 2009)
- flag: The query (lyx). If there is no query, then this value is NO_QUERY.
- user: the user that tweeted (robotickilldozr)
- text: the text of the tweet (Lyx is cool)

```python
from sklearn.model_selection import train_test_split

train_data, temp = train_test_split(raw_data, test_size=0.2, random_state=42)

val_data, test_data = train_test_split(temp, test_size=0.5, random_state=42)
```

```python
train_data.to_csv('../data/raw_data/train_data.csv', index=False, encoding='utf-8')
val_data.to_csv('../data/raw_data/val_data.csv', index=False, encoding='utf-8')
test_data.to_csv('../data/raw_data/test_data.csv', index=False, encoding='utf-8')
```

```python
train_part1 = train_data.iloc[:len(train_data)//2]
train_part2 = train_data.iloc[len(train_data)//2:]
```

```python
train_part1.to_csv('../data/raw_data/train_part1.csv', index=False, encoding='utf-8')
train_part2.to_csv('../data/raw_data/train_part2.csv', index=False, encoding='utf-8')
```
